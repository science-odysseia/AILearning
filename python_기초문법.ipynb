비교연산자

|비교연산자|의미|
|:---:|:---:|
|a `==` b|같다|
|a `!=` b|다르다|
|a `<` b|a는 b보다 작다|
|a `<=` b|a는 b보다 작거나 같다|
|a `>` b|a는 b보다 크다|
|a `>=` b|a는 b보다 크거나 같다|

````python
print(1 < 2)   # 참
print(3 > 5)   # 거짓
print(4 == 4)  # 참
print(5 != 5)  # 거짓
print(3 <= 3)  # 참
print(7 >= 10) # 거짓
````

논리연산자 `and`, `or`, `not`

`and` : 둘 다 True여야만 True, 그 외엔 False

`or` : 둘 다 False여야만 False, 그 외엔 True

`not` : True를 False로, False를 True로 변환

````python
print(True and True)     # 참
print(True and False)    # 거짓
print(False and True)    # 거짓
print(False and False)   # 거짓
print('-------------------')
print(True or True)      # 참
print(True or False)     # 참
print(False or True)     # 참
print(False or False)    # 거짓
print('-------------------')
print(not True)     # 거짓
print(not False)    # 참
````

조건문 if

if 문이 True이면 해당 조건문 실행.

False이면 해당 조건문을 실행하지 않고 넘김

````python
score = 80
if score >= 60:
    print("60점 이상입니다.")
if score >= 40:
    print("40점 이상입니다.")
if score >= 90:
    print("90점 이상입니다.")
````

if-else 구조

if 문이 False 일 때 else 실행.

따라서 반드시 단독으로 못 쓰이고 if문과 같이 쓰여야함.

````python
score = 80
if score >= 90:
    print("A학점입니다.")
else:
    print("A학점이 아닙니다.")
````

elif : else와 if의 합친 형태. else 문 안에 if가 들어간 구조랑 똑같다고 보면 된다.

아래처럼 if-elif-else 구조로 쓰임.

````python
score = 85
if score >= 90:
    print("A학점입니다.")
elif score >= 80:
    print("B학점입니다.")
else:
    print("C학점입니다.")
````

리스트와 딕셔너리

````python
list1 = [1, 2, 3, 4, 5]                                         # 리스트 예시 1
list2 = ["a", "b", "c", "d", "e"]                               # 리스트 예시 2
list3 = [1, "a", 3.14, True]                                    # 리스트 예시 3(혼합형)

dic1 = {"name": "Alice", "age": 30, "city": "New York"}         # 딕셔너리 예시 1
dic2 = {"fruit": "apple", "color": "red", "price": 1.5}         # 딕셔너리 예시 2
dic3 = {"number": 10, "string": "Hello", "boolean": False}      # 딕셔너리 예시 3(혼합형)
````

````python
print(list1[4])
print(list2[2])
print(list3[3])

print(dic1["name"])
print(dic2["color"])
print(dic3["boolean"])
````
